### Reading all the 3 datasets and retriving the ground truth trials

```python
import pandas as pd
import json
```

```python
sigir = pd.read_csv(r"dataset\sigir\qrels\test.tsv", sep="\t")
sigir
```

```python
trec21 = pd.read_csv(r"dataset\trec_2021\qrels\test.tsv", sep="\t")
trec21
```

```python
trec22 = pd.read_csv(r"dataset\trec_2022\qrels\test.tsv", sep="\t")
trec22
```

```python
all_dataset_cases= pd.concat([sigir, trec21, trec22], axis=0)
all_dataset_cases
```

##### The 14 pan cancer cases, existing within the 3 datasets

```python
cancer_cases = ["trec-20211", "trec-202111", "trec-202115", "trec-202120", "trec-202125", "trec-202127", "trec-202131", "trec-202134", "trec-202156", "trec-202161", "trec-202164", "trec-202170", "trec-202211", "sigir-201427"]
```

```python
# filtering the nct ids of the above cases to add to the corpus
cancer_cases_filtered = all_dataset_cases[all_dataset_cases["query-id"].isin(cancer_cases)]
cancer_cases_filtered
```

```python
cancer_cases_filtered.to_csv(r"dataset\GS_data\CTM_evaluation\pan_cancer_cases_GT_ncts.csv", index=None)
```

```python
nct_ids_for_pan_cancer_cases = list(set(cancer_cases_filtered["corpus-id"])) #incase the duplicates are retrived 
len(nct_ids_for_pan_cancer_cases)
```

### Checking the existance of the trials in the trial_info json, and if yes appending the elements to the gs_data having json of active+inactive trials. 

```python
with open("dataset/trial_info.json", 'r') as file:
    trial_corpus_json = json.load(file)
trial_corpus_json
```

```python
pan_cancer_14_cases_ncts_json = {}
for key in nct_ids_for_pan_cancer_cases:
    if key in trial_corpus_json:
        pan_cancer_14_cases_ncts_json[key] = trial_corpus_json[key]
    else:
        print(key, "not present")

len(pan_cancer_14_cases_ncts_json) #original length 5624, If we dont include the 2 trials not present it wont make much of difference as we they are classified as exluded for trec-202115
```

##### appending the trials for 14 GT pan cancer cases to the GS corpus

```python
with open("dataset\GS_data\GS_trials_info.json", 'r') as file:
    GS_corpus_json = json.load(file)
GS_corpus_json
```

#### Checking if GS_corpus already have the items of pan_cancer_14_cases_ncts_json

```python
GS_corpus_with_gt_cases_trials = GS_corpus_json.copy()
for key in pan_cancer_14_cases_ncts_json:
    if key in GS_corpus_json:
        # GS_corpus_with_gt_cases_trials[key].update(pan_cancer_14_cases_ncts_json[key])
        print(key,"present in GS_corpus")
    else:
        GS_corpus_with_gt_cases_trials[key] = pan_cancer_14_cases_ncts_json[key]
        print(key, "added to GS_corpus")

## only 15 were present in GS_corpus
```

```python
GS_corpus_with_gt_cases_trials
```

```python
with open (r"dataset\GS_data\CTM_evaluation\GS_corpus_with_gt_pan_cancer_cases_trials.json", "w") as f:
    json.dump(GS_corpus_with_gt_cases_trials, f, indent=4)
```
